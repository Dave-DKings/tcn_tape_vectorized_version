# Asset Universe Robustness Plan

This notebook defines the asset-universe robustness suite to evaluate after the final `Run21` training completes.

The goal is to test whether the agent's behavior is specific to the canonical training universe or whether it generalizes to materially different cross-sectional structures.


## 1. Why This Exists

The publication claim should not rest on one hand-curated 10-asset basket only.

A stronger evaluation story is:
- in-universe performance on the canonical 10
- adjacent-universe generalization on a standard mega-cap set
- stress testing on a concentrated tech-only set
- contrast testing on a defensive low-beta set

These universes are not random alternatives. Each one probes a different failure mode.


```python
from pprint import pprint

ASSET_UNIVERSE_PLAN = {
    "canonical_10": {
        "tickers": ["NVDA", "MSFT", "AMZN", "JPM", "BRK-B", "CAT", "XOM", "NEE", "GLD", "KO"],
        "purpose": "Primary in-universe benchmark and direct continuation of Run20/Run21 lineage.",
        "tests": [
            "Baseline policy quality on the trained universe",
            "Whether the final Run21 curriculum produces the strongest deployable checkpoint on its native basket",
            "Phase-by-phase comparison against Run20"
        ],
        "expected_characteristics": [
            "Balanced sector mix",
            "Includes cyclicals, defensives, and a hedge asset",
            "Best direct comparability with prior runs"
        ],
    },
    "top10_market_cap": {
        "tickers": ["MSFT", "NVDA", "AAPL", "AMZN", "GOOGL", "META", "BRK-B", "TSLA", "LLY", "JPM"],
        "purpose": "Objective mega-cap generalization test.",
        "tests": [
            "Whether the agent depends on the hand-curated canonical basket",
            "How well it handles a more standardized benchmark-style universe",
            "Whether alpha survives under higher average correlation and mega-cap concentration"
        ],
        "expected_characteristics": [
            "More tech-heavy than the canonical 10",
            "Less natural diversification from commodities and gold",
            "Harder to beat equal-weight if cross-sectional dispersion is narrower"
        ],
    },
    "tech_only_10": {
        "tickers": ["MSFT", "NVDA", "AAPL", "AMZN", "GOOGL", "META", "TSLA", "AMD", "NFLX", "AVGO"],
        "purpose": "High-beta concentration stress test.",
        "tests": [
            "Whether the policy still works when hedging options are limited",
            "Whether the risk and discipline experts matter in a high-correlation universe",
            "Whether turnover and drawdown explode under growth concentration"
        ],
        "expected_characteristics": [
            "High internal correlation",
            "High momentum and crash sensitivity",
            "Likely the hardest universe for churn control"
        ],
    },
    "defensive_10": {
        "tickers": ["KO", "PG", "JNJ", "PEP", "WMT", "COST", "BRK-B", "NEE", "DUK", "GLD"],
        "purpose": "Low-beta contrast universe.",
        "tests": [
            "Whether the agent still adds value when market beta is lower",
            "Whether it can extract signal from a less explosive cross-section",
            "Whether turnover naturally compresses in a defensive basket"
        ],
        "expected_characteristics": [
            "Lower volatility than canonical or tech-only baskets",
            "Less explosive upside",
            "Useful contrast against the high-beta universes"
        ],
    },
}

pprint(ASSET_UNIVERSE_PLAN)
```

## 2. What Each Universe Tests

### Canonical 10
Use this as the primary benchmark. It answers: did the final training run produce a better native-universe policy than prior runs?

### Top-10 Market-Cap
This is the most defensible alternate universe because it is objective rather than hand-curated. It tests whether performance survives on a standard mega-cap equity universe.

### Tech-Only 10
This is a concentration stress test. If the agent depends too much on cross-sector diversification, this universe will expose it quickly.

### Defensive 10
This is the opposite of tech-only. It tests whether the agent has genuine policy skill when the universe is less explosive and less momentum-driven.


```python
import pandas as pd

rows = []
for name, spec in ASSET_UNIVERSE_PLAN.items():
    rows.append({
        "universe": name,
        "size": len(spec["tickers"]),
        "purpose": spec["purpose"],
        "primary_risk_tested": spec["tests"][0],
        "example_tickers": ", ".join(spec["tickers"][:5]),
    })

summary_df = pd.DataFrame(rows)
summary_df
```

## 3. Recommended Evaluation Order

Use this order after `Run21` finishes:

1. `canonical_10`
2. `top10_market_cap`
3. `tech_only_10`
4. `defensive_10`

Reason:
- start with the native universe
- then move to the most publication-defensible generalization test
- then run the concentrated stress test
- then finish with the low-beta contrast test


## 4. Publication Framing

A clean writeup can position these universes as follows:

- `canonical_10`: in-universe performance
- `top10_market_cap`: objective alternate-universe generalization
- `tech_only_10`: concentration and correlation stress test
- `defensive_10`: low-beta robustness and lower-dispersion contrast test

That gives a stronger story than evaluating only one universe or a set of ad hoc baskets.
