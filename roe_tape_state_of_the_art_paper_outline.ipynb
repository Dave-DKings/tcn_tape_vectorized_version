## 1. Paper Thesis and Scope

### 1.1 Core thesis
The paper should argue that **portfolio RL becomes more stable, interpretable, and regime-robust when multi-objective control is made explicit in the architecture** rather than buried inside a monolithic actor-critic and scalar reward.

### 1.2 Main method identity
The final method is **ROE-TAPE**, not the earlier simpler `TAPE-TCN` framing.

### 1.3 Central contributions to emphasize
1. **Structured multi-objective portfolio RL** via routed objective experts.
2. **Separate actor and critic specialization** for return, risk, and discipline.
3. **Regime-conditioned routing** over expert policies.
4. **Simplex-stable Dirichlet policy engineering** for valid portfolio weights.
5. **A serious evaluation protocol**: deterministic vs stochastic, benchmark-relative, horizon-aware, regime-aware, and CVaR-aware.

### 1.4 Important non-goal
Do not try to make the paper about every module equally. The paper needs a center of gravity. The center should be **ROE-TAPE as a structured multi-objective portfolio RL framework**.

## 6. Recommended Introduction Outline

### 6.1 Problem motivation
- Classical portfolio optimization offers structure but relies on fragile covariance assumptions.
- End-to-end deep RL offers adaptivity but often fails on turnover control, risk discipline, and regime shifts.
- Portfolio RL is a multi-objective problem even when implemented as a scalar reward.

### 6.2 Gap in prior work
- Many finance RL systems optimize blended objectives inside a monolithic policy.
- Many papers report limited benchmark comparisons or a single evaluation horizon.
- Many papers use action parameterizations that require post-hoc clipping or weak portfolio-validity handling.

### 6.3 Proposed solution
- ROE-TAPE separates objectives explicitly.
- The model routes across expert behaviors by regime/state.
- The action distribution is simplex-native and numerically stabilized.
- The evaluation is benchmark-aware, horizon-aware, and regime-aware.

## 13. Claims, Non-Claims, and Threats to Validity

### 13.1 Strong claims you can make if the evidence holds
- Explicit objective separation improves portfolio RL stability and benchmark-relative performance.
- Regime-conditioned routing improves robustness across evaluation windows.
- Simplex-stable Dirichlet engineering is necessary for stable portfolio RL on constrained allocation spaces.
- A horizon-aware, regime-aware evaluation protocol reveals strengths and weaknesses hidden by single-period backtests.

### 13.2 Claims you should avoid
- universal state of the art across all portfolio RL tasks
- universal superiority across all universes and all market conditions
- causal claims about modules that were not directly ablated

### 13.3 Threats to validity
- too many simultaneous changes
- universe-specific overfitting
- benchmark dependence on universe composition
- limited external baselines
- no live execution results
- evaluation-date selection bias if windows are narrowed post hoc
