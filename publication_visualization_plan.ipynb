## 6) Figure Design Notes

A few concrete design rules for the final paper figures:
- put curriculum transitions as vertical markers on training plots
- use the same color mapping everywhere for experts:
  - return = blue
  - risk = orange
  - discipline = green
- use the same color mapping everywhere for baselines:
  - agent = dark navy
  - equal-weight = gray
  - SPY = black
- avoid plotting too many checkpoints in one figure unless the figure is explicitly a heatmap
- use medians and interquartile ranges for stochastic summaries where possible


```python
FIGURE_STYLE_GUIDE = {
    "expert_colors": {
        "return": "#1f77b4",
        "risk": "#ff7f0e",
        "discipline": "#2ca02c",
    },
    "baseline_colors": {
        "agent": "#0b3c5d",
        "equal_weight": "#7f7f7f",
        "spy": "#111111",
    },
    "notes": [
        "Use phase markers on training curves.",
        "Use violin or box plots for stochastic distributions.",
        "Use heatmaps for checkpoint x horizon tables.",
        "Prefer median and IQR for robustness plots."
    ]
}

FIGURE_STYLE_GUIDE
```
